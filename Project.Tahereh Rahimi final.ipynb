### Intital Setup

Set up the foundational structure for your weekly task and habit tracker.

In this tracker, a **task** is something you need to do once, on a specific day (e.g., "Go to supermarket" on Monday), while a **habit** is something you want to do every day of the week (e.g., "Drink Water"). That's why we store them in two different dictionaries.

#### Instructions:
1. Create a list containing all the days of the week **('Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday')**.
2. Initialize an empty dictionary that will be used to track habits for the week.
3. Create a second dictionary that uses the days of the week as keys, with each key initially linked to an empty dictionary for storing tasks.

**Tip:** An empty dictionary is written as `{}`. For step 3, you can build a dictionary from a list in one line with a *dictionary comprehension*. For example, `{size: 0 for size in ['S', 'M', 'L']}` gives `{'S': 0, 'M': 0, 'L': 0}`.

#### What your data will look like
Once a few tasks and habits have been added (in Part 2), your two dictionaries will look something like this:

```python
# Tasks dictionary: each day has its own dictionary of tasks (True = completed, False = not completed)
{'Monday': {'Go to supermarket': False, 'Pay bills': True}, 'Tuesday': {}, 'Wednesday': {}, ...}

# Habits dictionary: each habit has its own dictionary of days (True = completed, False = not completed)
{'Drink Water': {'Monday': True, 'Tuesday': False, 'Wednesday': False, ...}}
```

Keep this structure in mind, since every part of the project uses it.

````python
# Create a list of days
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

# Create an empty dictionary for habits
habits = {}

# Create a dictionary for tasks, with an empty dictionary for each day
tasks = {day: {} for day in days}

print(days)
print(habits)
print(tasks)
````

### Add Tasks and Habits
Build functions that let users add tasks to specific days and habits for the entire week.
#### Instructions
1. **Create a function to add tasks:**
- Ask the user which day they want to add the task to.
- Make sure the day is valid (it must be one of the days in your list). If it isn't, display a message saying the day is invalid.
- Ask for the task name.
- Store the task in your tasks dictionary, under the correct day.
- Set the task’s value to *False* to show it hasn’t been completed yet.


2. **Create a function to add habits:**
- Ask the user for the name of the habit (e.g., “Drink Water”).
- Check if the habit already exists. If it does, display a message saying so.
- If it’s new, add it to your habits dictionary.
- For that habit, create a nested dictionary with all 7 days set to *False*.


#### Tips:
- Use `.strip()` to remove extra spaces and `.title()` to capitalize each word. For example, `"  monDAY ".strip().title()` gives `"Monday"`.
- Use `in` to check if a value is in a list (or a key is in a dictionary), and `not in` for the opposite. For example, `'Apple' in ['Apple', 'Pear']` is `True`.
- To reach a dictionary inside a dictionary, use two sets of square brackets. For example, `pets['Rex']['Walk'] = False`.
- Your functions can change the dictionaries from Part 1 directly; you don't need to pass them as arguments.
- Writing `return` on its own stops a function right away (useful after showing an error message).
- **If the day does not exist, show a helpful error message.**

**Testing your functions:** Defining a function does not run it, so call it in a cell (for example, `add_task()`) to test it. When you're done, comment out the call (add a # in front of it) so it doesn't ask for input every time you run the notebook. If you re-run the Part 1 cell, your dictionaries will be reset to empty.

````python
# Function to add a task to a specific day
def add_task():
    day = input("Enter the day: ").strip().title()

    # Check whether the entered day is valid
    if day not in days:
        print("Invalid day. Please enter a valid day.")
        return

    task_name = input("Enter the task name: ").strip().title()

    # Check whether the task name is empty
    if task_name == "":
        print("Task name cannot be empty.")
        return

    # Check whether the task already exists for this day
    if task_name in tasks[day]:
        print("This task has already been added for this day.")
        return

    # Add the task with False as its initial status
    tasks[day][task_name] = False

    print(task_name, "added successfully for", day)


# Function to add a habit for the entire week
def add_habit():
    habit_name = input("Enter the habit name: ").strip().title()

    # Check whether the habit name is empty
    if habit_name == "":
        print("Habit name cannot be empty.")
        return

    # Check whether the habit already exists
    if habit_name in habits:
        print("This habit already exists.")
        return

    # Add the habit with all seven days set to False
    habits[habit_name] = {day: False for day in days}

    print(habit_name, "added successfully.")
````

### Marking Tasks and Habits as Complete
In this part, you'll implement two functions to allow users to mark their progress: one for tasks and one for habits.
#### Instructions:
1. **Create a function to mark tasks as complete:**
- Display all incomplete tasks along with the day they belong to.
- If there are no incomplete tasks, display a message saying so (for example, "All tasks completed!").
- Ask the user to enter the name of the task they completed.
- Check if the task exists. If it doesn't, display a message saying the task was not found.
- If it exists, update the task in your tasks dictionary by setting its value to *True*.
- **BONUS:** If the task is from a day before today, print an encouraging or funny message like: "Task marked as complete. Better late than never!"
    - **Tip:** You can use the *datetime* library (see the tips below).

2. **Create a function to mark habits as complete:**
- Ask the user to enter the day and the name of the habit.
- Make sure the day is valid and check if the habit exists. Display a message if the day is invalid or if the habit does not exist.
- Update the habit’s status for that day by setting its value to *True*.

#### Tips:
- Use **.strip()** and **.title()** to clean up user input.
- Tasks are stored inside each day, so use a loop inside a loop: one over the days, and one over that day's tasks (see the helpful note below for `.items()`).
- `not` flips `True` and `False`, so `if not done:` runs when `done` is `False`.
- To know if you found something, start with a variable like `found = False`, set it to `True` when you find it, and check it after the loop.
- **BONUS:** After `import datetime`, `datetime.datetime.now().strftime("%A")` gives today's day name (for example, `'Monday'`), and `days.index('Monday')` gives its position in your list (`0`). A smaller position means an earlier day.
- **If the day, habit or task doesn't exist, show a helpful error message.**
  
**Helpful note:** You can access a dictionary’s information using loops like this:

This will print each key and its value.

You can also loop through just the keys with `for key in my_dict:` or just the values with `for value in my_dict.values():`

````python
 # Function to mark a task as complete
def mark_task_complete():
    print("Incomplete Tasks:")

    found_tasks = False

    # Display all incomplete tasks and their days
    for day in days:
        for task_name, status in tasks[day].items():
            if status == False:
                print(day, ":", task_name)
                found_tasks = True

    # Check whether there are any incomplete tasks
    if found_tasks == False:
        print("All tasks completed!")
        return

    selected_task = input(
        "Enter the task name to mark as complete: "
    ).strip().title()

    found = False
    already_completed = False

    # Search for the selected task
    for day in days:
        for task_name, status in tasks[day].items():

            if task_name == selected_task:
                found = True

                if status == False:
                    tasks[day][task_name] = True
                    print("Task marked as complete.")
                    return

                else:
                    already_completed = True

    # Display the appropriate message
    if already_completed:
        print("This task has already been completed.")
    elif found == False:
        print("Task not found.")


# Function to mark a habit as complete
def mark_habit_complete():
    day = input("Enter the day: ").strip().title()

    # Check whether the day is valid
    if day not in days:
        print("Invalid day. Please enter a valid day.")
        return

    habit_name = input("Enter the habit name: ").strip().title()

    # Check whether the habit exists
    if habit_name not in habits:
        print("Habit not found.")
        return

    # Check whether the habit is already completed for that day
    if habits[habit_name][day] == True:
        print("This habit has already been completed for this day.")
        return

    # Update the habit status
    habits[habit_name][day] = True

    print(habit_name, "marked as complete for", day)
````

````python
mark_task_complete()
mark_habit_complete()
````

### Removing Tasks and Habits

#### Instructions:
1. **Create a function to remove a task:**
- Ask the user which day the task is on.
- Make sure the day is valid. If it isn't, display a message saying the day is invalid.
- Show the list of tasks for that day. If there are none, display a message saying so.
- Ask the user which task to remove.
- If the task exists, delete it from the dictionary using del and display a message saying which task was deleted. If it doesn't exist, display a message saying so.

2. **Create a function to remove a habit:**
- If no habits are being tracked, display a message saying so.
- Display all habits that are currently being tracked.
- Ask the user which habit to remove.
- If it exists in the dictionary, delete it using del and display a message saying which habit was deleted. If it doesn't exist, display a message saying so.

#### Tips:
- Use **.strip()** and **.title()** to clean up user input.
- An empty dictionary counts as `False`, so `if not my_dict:` runs when the dictionary is empty. Use this to check if there's data before accessing or deleting something.
- `del` works with two sets of square brackets too. For example, `del pets['Rex']['Walk']` removes `'Walk'` from Rex's dictionary.
- **If the day, habit or task doesn't exist, show a helpful error message.**

````python
# Remove a task from a selected day
def remove_task():
    day = input("Enter the day: ").strip().title()

    # Check whether the entered day is valid
    if day not in days:
        print("Invalid day. Enter a valid day.")
        return

    # Check whether there are any tasks for this day
    if not tasks[day]:
        print("There are no tasks for this day.")
        return

    # Display the tasks scheduled for this day
    print("Tasks for", day, ":")
    for task_name in tasks[day]:
        print("-", task_name)

    # Ask which task should be removed
    task_name = input("Enter the task name you want to remove: ").strip().title()

    # Remove the task if it exists
    if task_name in tasks[day]:
        del tasks[day][task_name]
        print("Task", task_name, "was removed successfully from", day)
    else:
        print("Task not found for this day.")
````

````python
remove_task()
````

````python
# Remove a habit from the tracker
def remove_habit():

    # Check whether any habits are being tracked
    if not habits:
        print("There are no habits being tracked.")
        return

    # Display all tracked habits
    print("Tracked habits:")
    for habit_name in habits:
        print("-", habit_name)

    # Ask which habit should be removed
    habit_name = input("Enter the habit name you want to remove: ").strip().title()

    # Remove the habit if it exists
    if habit_name in habits:
        del habits[habit_name]
        print("Habit", habit_name, "was removed successfully.")
    else:
        print("Habit not found.")
````

### Generating Weekly and Daily Reports

#### Instructions:
1. **Create a weekly report function**

This function should summarize the user's progress over the entire week.
- For habits:
    - If there are no habits, display a message saying so.
    - Loop through each habit in your habits dictionary.
    - Count how many days the habit was marked as complete, using a loop (for example, with a counter that starts at 0) or `sum()` (see the tips below).
    - Display the habit name and the number of days it was completed out of 7 (for example, "Drink Water: 3/7 days complete").
- For tasks:
    - Loop through all days and collect tasks into two separate lists:
        - One for completed tasks
        - One for not completed tasks
    - For each task, include the task name and its day in parentheses (for example, "Go to supermarket (Monday)").
    - Display both lists clearly.
    - If there are no tasks, display a message saying so.

2. **Create a daily report function**

This function allows the user to view all their activity for a specific day.

- Ask the user to enter a day (e.g., "Monday").
- If the day is valid:
    - Display all tasks for that day, showing whether each one is complete or not.
    - Display all habits, showing whether they were completed on that day.
    - If there are no tasks or no habits, display a message saying so.
- Use ✅ for completed and ❌ for not completed.

#### Tips:
- Use **.strip()** and **.title()** to make sure the input day matches the format in your list.
- Use clear formatting in your print statements to separate sections.
- Empty lists and dictionaries count as `False`, so `if not my_list:` runs when the list is empty.
- `sum()` can count how many items match a condition. For example, `sum(1 for t in [18, 25, 30] if t > 20)` gives `2` (it adds 1 for every temperature above 20).
- You can choose between two values in one line, for example: `label = "Yes" if bought else "No"`. Copy ✅ and ❌ from these instructions and paste them into your strings.
- **If the day doesn't exist, show a helpful error message.**

````python
# Generate a report for a selected day
def daily_report():

    day = input("Enter the day for the report: ").strip().title()

    # Validate the entered day
    if day not in days:
        print("Invalid day. Enter a valid day.")
        return

    print("\nDaily Report:", day)

    # Display tasks for the selected day
    print("\nTasks:")

    if not tasks[day]:
        print("No tasks found for this day.")
    else:
        for task_name, status in tasks[day].items():

            # Display the appropriate completion symbol
            if status == True:
                print(task_name, ": ✅")
            else:
                print(task_name, ": ❌")

    # Display habit status for the selected day
    print("\nHabits:")

    if not habits:
        print("No habits found.")
    else:
        for habit_name, status in habits.items():

            if status[day] == True:
                print(habit_name, ": ✅")
            else:
                print(habit_name, ": ❌")

    print("\nEnd of daily report")
````

````python
daily_report()
````

````python
# Generate a report for the entire week
def weekly_report():

    print("\nWeekly Report")

    # Create two empty lists for completed and incomplete tasks
    completed_tasks = []
    incomplete_tasks = []

    # Go through each day and collect the tasks
    for day in days:

        for task_name, status in tasks[day].items():

            if status == True:
                completed_tasks.append(task_name + " (" + day + ")")
            else:
                incomplete_tasks.append(task_name + " (" + day + ")")

    # Display completed tasks
    print("\nCompleted Tasks:")

    if completed_tasks:
        for task in completed_tasks:
            print("✅", task)
    else:
        print("No completed tasks.")

    # Display incomplete tasks
    print("\nIncomplete Tasks:")

    if incomplete_tasks:
        for task in incomplete_tasks:
            print("❌", task)
    else:
        print("No incomplete tasks.")

    # Display weekly habit progress
    print("\nWeekly Habits:")

    if not habits:
        print("No habits found.")

    else:
        for habit_name, status in habits.items():

            completed_days = 0

            # Count how many days the habit was completed
            for day in days:
                if status[day] == True:
                    completed_days += 1

            print(habit_name, ":", completed_days, "/7 days complete")

    print("\nEnd of Weekly Report")
````

### Building the Main Menu System
In this final part, you'll create a main control loop that acts as the user interface for your program. This allows users to interact with your habit and task tracker through a series of text-based menu options.

#### Instructions:
1. **Set up a loop:** Use a loop to keep showing the menu until the user chooses to exit (see the tips below).

2. **Display the main menu:** Show numbered options for:
    1. Adding a Task or Habit
    2. Marking a Task or Habit as complete
    3. Removing a Task or Habit
    4. Viewing reports
    5. Exiting the program

3. **Handle user input:** Ask the user to choose an option using input(). Based on the number they enter:
- Option 1: Ask whether they want to add a Task or a Habit. Call the correct function.
- Option 2: Ask whether they want to mark a Task or Habit as complete. Call the right function.
- Option 3: Ask whether they want to remove a Task or a Habit. Call the appropriate function.
- Option 4: Ask whether they want a Weekly or Daily report. Call the corresponding function.
- Option 5: Exit the loop using break.

4. **Validate input:** Check for valid input at each step. If the user enters an invalid choice, show an error message and return to the menu.

#### Tips:
- `while True:` keeps a loop running until it reaches `break`.
- Put your menu loop inside a function (for example, `main()`) and call it at the end of the cell.
- `input()` always returns a string, so compare the user's choice to `'1'`, `'2'`, etc. (with quotes).
- `.lower()` turns text into lowercase, so the user can type `A` or `a`.
- Run all the cells above (Parts 1 to 5) before running your menu. You can use *Run → Run All Cells*.

````python
# Main menu to control the Weekly Task and Habit Tracker
def main():

    while True:

        print("\n===== Weekly Task & Habit Tracker =====")
        print("1. Add a Task or Habit")
        print("2. Mark a Task or Habit as Complete")
        print("3. Remove a Task or Habit")
        print("4. View Reports")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        # Option 1: Add a task or habit
        if choice == "1":

            selection = input("Do you want to add a Task or Habit? ").strip().lower()

            if selection == "task":
                add_task()
            elif selection == "habit":
                add_habit()
            else:
                print("Invalid choice. Please enter Task or Habit.")

        # Option 2: Mark a task or habit as complete
        elif choice == "2":

            selection = input("Do you want to mark a Task or Habit as complete? ").strip().lower()

            if selection == "task":
                mark_task_complete()
            elif selection == "habit":
                mark_habit_complete()
            else:
                print("Invalid choice. Please enter Task or Habit.")

        # Option 3: Remove a task or habit
        elif choice == "3":

            selection = input("Do you want to remove a Task or Habit? ").strip().lower()

            if selection == "task":
                remove_task()
            elif selection == "habit":
                remove_habit()
            else:
                print("Invalid choice. Please enter Task or Habit.")

        # Option 4: View weekly or daily reports
        elif choice == "4":

            selection = input("Do you want a Weekly or Daily report? ").strip().lower()

            if selection == "weekly":
                weekly_report()
            elif selection == "daily":
                daily_report()
            else:
                print("Invalid choice. Please enter Weekly or Daily.")

        # Option 5: Exit the program
        elif choice == "5":
            print("Thank you for using the Weekly Task & Habit Tracker!")
            break

        # Handle invalid menu choices
        else:
            print("Invalid option. Please enter a number between 1 and 5.")


# Run the main program
main()
````
